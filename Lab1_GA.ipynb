```python
import random

# Grid size
ROWS = 10
COLS = 10

# Start and goal positions
START = (0, 0)
GOAL = (9, 9)

# Obstacles
OBSTACLES = {
    (1, 2), (2, 2), (3, 2),
    (3, 4), (4, 4), (5, 4),
    (5, 6), (6, 6), (7, 6),
    (7, 8), (8, 8)
}

# Genetic Algorithm parameters
POPULATION_SIZE = 50
CHROMOSOME_LENGTH = 25
MUTATION_RATE = 0.1
CROSSOVER_RATE = 0.8
GENERATIONS = 100

# Possible movements
MOVES = ['U', 'D', 'L', 'R']


# Move the robot
def move(position, direction):
    row, col = position

    if direction == 'U':
        row -= 1
    elif direction == 'D':
        row += 1
    elif direction == 'L':
        col -= 1
    elif direction == 'R':
        col += 1

    return row, col


# Evaluate one path
def evaluate(path):
    position = START

    distance = 0
    collisions = 0
    turns = 0

    previous_move = None

    for direction in path:

        new_position = move(position, direction)

        # Check boundary
        if (new_position[0] < 0 or new_position[0] >= ROWS or
                new_position[1] < 0 or new_position[1] >= COLS):

            collisions += 1
        # Check obstacle
        elif new_position in OBSTACLES:

            collisions += 1

        else:
            position = new_position

        distance += 1

        # Count turns
        if previous_move is not None and previous_move != direction:
            turns += 1

        previous_move = direction

    # Distance from final position to goal
    goal_distance = abs(position[0] - GOAL[0]) + \
                    abs(position[1] - GOAL[1])

    # Fitness function
    fitness = (
        distance +
        (100 * collisions) +
        (10 * goal_distance) +
        (2 * turns)
    )

    return fitness, position, distance, collisions, turns


# Create initial population
def create_population():
    population = []

    for i in range(POPULATION_SIZE):
        chromosome = []

        for j in range(CHROMOSOME_LENGTH):
            chromosome.append(random.choice(MOVES))

        population.append(chromosome)

    return population


# Selection using tournament selection
def selection(population):
    tournament = random.sample(population, 3)

    best = tournament[0]

    for individual in tournament:
        if evaluate(individual)[0] < evaluate(best)[0]:
            best = individual

    return best


# Crossover
def crossover(parent1, parent2):

    if random.random() > CROSSOVER_RATE:
        return parent1[:]

    point = random.randint(1, CHROMOSOME_LENGTH - 1)

    child = parent1[:point] + parent2[point:]

    return child


# Mutation
def mutation(path):

    for i in range(CHROMOSOME_LENGTH):

        if random.random() < MUTATION_RATE:
            path[i] = random.choice(MOVES)

    return path


# Genetic Algorithm
def genetic_algorithm():

    population = create_population()

    best_path = None
    best_fitness = float('inf')

    for generation in range(GENERATIONS):

        # Find best individual
        for individual in population:

            fitness = evaluate(individual)[0]

            if fitness < best_fitness:
                best_fitness = fitness
                best_path = individual[:]

        # Create new population
        new_population = []

        while len(new_population) < POPULATION_SIZE:

            parent1 = selection(population)
            parent2 = selection(population)

            child = crossover(parent1, parent2)

            child = mutation(child)

            new_population.append(child)

        population = new_population

        # Display progress
        if generation % 10 == 0:
            print("Generation:", generation,
                  "Best Fitness:", best_fitness)

    return best_path, best_fitness


# Run Genetic Algorithm
best_path, best_fitness = genetic_algorithm()


# Display result
fitness, final_position, distance, collisions, turns = evaluate(best_path)

print("\n----- FINAL RESULT -----")

print("Best Path:")
print(best_path)

print("Fitness:", fitness)
print("Final Position:", final_position)
print("Distance:", distance)
print("Collisions:", collisions)
print("Turns:", turns)

if final_position == GOAL:
    print("Goal Reached!")
else:
    print("Goal Not Reached")
#https://ieeexplore.ieee.org/document/9131750
```
