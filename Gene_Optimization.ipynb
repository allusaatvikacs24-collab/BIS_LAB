```python

import random

random.seed(42)

# Generate dataset
X = [[i / 100, random.random()] for i in range(100)]
y = [1 if row[0] > 0.5 else 0 for row in X]

# Initialize chromosomes: (feature, threshold)
population = [(0, 0.5)] + [
    (random.randint(0, 1), random.random())
    for _ in range(29)
]

# Fitness evaluation
def fitness(chromosome):
    feature, threshold = chromosome
    predictions = [
        1 if row[feature] > threshold else 0
        for row in X
    ]
    return sum(a == b for a, b in zip(predictions, y)) / len(y)

# Gene Expression Algorithm
for generation in range(20):

    population.sort(key=fitness, reverse=True)

    # Selection
    new_population = population[:5]

    while len(new_population) < 30:

        p1, p2 = random.sample(population[:15], 2)

        # Crossover
        child = (p1[0], p2[1])

        # Mutation
        if random.random() < 0.1:
            child = (random.randint(0, 1), random.random())

        new_population.append(child)

    population = new_population

# Best solution
best = max(population, key=fitness)

print("Best Classification Rule:")
print("IF X1 >", round(best[1], 2), "THEN Class 1 ELSE Class 0")
print("Best Accuracy:", round(fitness(best) * 100, 2), "%")
#https://ieeexplore.ieee.org/document/8285696?utm_source=chatgpt.com
```
